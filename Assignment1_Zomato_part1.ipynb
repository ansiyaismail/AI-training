```python
#identify the duplicate records
duplicate_count = data.duplicated().sum()
print(f"Total duplicate rows: {duplicate_count}")
```

```python
data["Average Cost for two"] = data["Average Cost for two"].fillna(0)

# 2. Convert to whole numbers (integer)
data["Average Cost for two"] = data["Average Cost for two"].astype('int64')

print(data["Average Cost for two"].dtype)
```

```python
average_cost = data["Average Cost for two"].mean()

# Print the result rounded to two decimal places
print(f"The average cost for two is: {average_cost:.2f}")
```

```python
#performing basics statistical analysis
numeric_summary = data.describe()
print("--- Numerical Columns Statistical Summary ---")
print(numeric_summary)
```

```python
# Generate summary stats for text/object columns
categorical_summary = data.describe(include='O')
print("\n--- Categorical Columns Statistical Summary ---")
print(categorical_summary)
```
